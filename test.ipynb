```python
from scipy.sparse import csr_matrix, load_npz, save_npz
import numpy as np
from pathlib import Path

DATA_DIR = Path ("data/tidied_360k")


def load_tidied_data() -> csr_matrix:
    sparse_matrix = load_npz(DATA_DIR / "user_artist_matrix.npz")
    assert isinstance(sparse_matrix, csr_matrix)
    return sparse_matrix

train = load_npz(DATA_DIR / "train.npz")
val = load_npz(DATA_DIR / "validation.npz")
test = load_npz(DATA_DIR / "test.npz")
```

```python
train_counts = np.asarray((train > 0).sum(axis=0)).ravel()   # listeners per artist in train

for name, M in [("val", val), ("test", test)]:
    items = np.unique(M.nonzero()[1])
    missing = items[train_counts[items] == 0]
    print(name, "artists absent from train:", len(missing))
```

```python
# turn sparse matrix into mask with binary if listened to artist
# then sum along columns, creating one row with total number of listens per artist
train_counts_matrix = (train > 0).sum(axis=0)

# then convert into a simple flat array
train_counts = np.asarray(train_counts_matrix).ravel()
```
